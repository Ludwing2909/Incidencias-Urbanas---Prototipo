```python
from ml.data import validate, near_pairs
frame = validate('data/demostracion.csv')
frame.groupby(['tipo_dato','particion','categoria']).size().to_frame('n')
```

```python
frame.assign(longitud=frame.texto_reclamo.str.len()).groupby('categoria').longitud.describe()
```

```python
pairs = near_pairs(frame)
print('Pares cercanos para inspección:', len(pairs))
print('Grupos:', frame.grupo_origen.nunique())
```
